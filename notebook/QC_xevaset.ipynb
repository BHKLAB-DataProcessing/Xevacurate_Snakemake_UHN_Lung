## 4. PDX Growth Cure

```r
# set color of lines, error visualization with errorbar
plotPDX(x.set, batch = "f4.PHLC191.TPX.palbociclib", 
        control.col = "#a6611a", 
        treatment.col = "#018571",
        major.line.size = 1,
        max.time = 100,
        SE.plot='errorbar')
```

```r
# normalize volume, change color of lines, error visualization with ribbon
plotPDX(x.set, batch = "f4.PHLC191.TPX.palbociclib", 
        vol.normal = TRUE, 
        control.col = "#a6611a", 
        treatment.col = "#018571",
        major.line.size = 1,
        max.time = 100,
        SE.plot='ribbon')
```
